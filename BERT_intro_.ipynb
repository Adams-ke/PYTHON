#### BERT Embeddings for Log Classification
####
#### This notebook demonstrates how to generate embeddings using a pre-trained BERT model
#### for log message classification. BERT (Bidirectional Encoder Representations from Transformers)
#### converts text logs into dense numerical vectors (embeddings) that capture semantic meaning.
####
#### These embeddings serve as input features for a downstream classifier to automatically
#### categorize log messages (e.g., security_breach, hardware_failure, system_error).
####
#### Key concepts covered:
#### - Tokenization: Converting raw text to token IDs with [CLS] and [SEP] special tokens
#### - Padding & Truncation: Handling variable-length inputs to a fixed max length
#### - Input IDs: Mapping tokens to BERT's vocabulary indices
#### - Attention Masks: Distinguishing real tokens from padding
#### - Token Type IDs: Differentiating between multiple sentences
#### - Pooled Output: Sentence-level embeddings for classification
#### - Sequence Output: Token-level embeddings for each position
####
#### The final pooled_output vectors (shape: batch_size, 768) are used as features
#### to train a classifier for automated log analysis.


```python

import tensorflow_hub as hub
import tensorflow_text as text
```

```python
import kagglehub

# Download latest version
path = kagglehub.model_download("tensorflow/bert/tensorFlow2/bert-en-uncased-l-8-h-768-a-12")

print("Path to model files:", path)
```

```python
import kagglehub

# Download latest version
preprocess_path = kagglehub.model_download("tensorflow/bert/tensorFlow2/en-uncased-preprocess")

print("Path to model files:", path)
```

```python
bert_preprocess_model = hub.KerasLayer(preprocess_path)
```

```python
text_test = ['nice movie indeed', 'I love python programming']
text_preprocessed = bert_preprocess_model(text_test)
text_preprocessed.keys()
```

```python
text_preprocessed['input_mask']
```

## Why 5 tokens instead of 3?
###The original sentence has 3 words, but BERT's tokenizer adds:
###1. [CLS] - Special token added to the beginning
###   2. [SEP] - Separator token added to the end
### This increases the token count from 3 to 5.

### The shape shows 218 tokens - this is the maximum sequence length.
### Shorter sentences are padded with [PAD] tokens to reach this length.
### Only the first few tokens contain actual words; the rest are padding.

```python
text_preprocessed['input_type_ids']
```

#### input_type_ids indicate which sentence each token belongs to.
#### This is useful for tasks involving multiple sentences (e.g.,
#### next sentence prediction, question answering, or duplicate detection).

```python
text_preprocessed['input_word_ids']
```

#### input_word_ids: Maps each token in the input to a unique integer ID
#### from BERT's pre-trained vocabulary. For example:
#### "hello world" -> [7592, 2088] (token IDs)
#### Special tokens like [CLS] and [SEP] also get their own IDs.

```python
bert_model = hub.KerasLayer(path)

bert_results = bert_model(text_preprocessed)
bert_results.keys()
```

```python
bert_results['pooled_output']
```

#### pooled_output: Shape (2, 768)
####  - Row 1: [0.1897, -0.8059, -0.2774, ..., 0.4507, -0.1045, -0.9656] for the first sentence
####   - Row 2: [-0.2286, 0.9856, -0.3307, ..., 0.2099, -0.0603, 0.2875] fro the 2nd sentence
#### Each row is a fixed-length embedding representing a complete sentence.
#### The classifier model will use these 768 features as input.

```python
bert_results['sequence_output'] #Individual word embedding vectors
```


#### Each sentence is represented by a 768-dimensional vector.
#### Even padding tokens (the 0s added to fill to max_length like 128) get embeddings,
#### since BERT is contextualized and computes representations for every token.
#### The [CLS] token aggregates information from the entire sequence,
#### and pooled_output is derived from this [CLS] token's final state.

#### Convenience and clarity:
#### - sequence_output: Explicit, obvious name for token-level embeddings
#### - encoder_outputs[-1]: Less obvious, requires knowing indexing
####
#### They keep both for:
#### 1. Backward compatibility (older code uses sequence_output)
#### 2. Flexibility (encoder_outputs lets you access intermediate layers)
